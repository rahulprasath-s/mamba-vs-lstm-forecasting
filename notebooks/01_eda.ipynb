# Exploratory Data Analysis

Use this notebook to inspect raw time-series CSVs before training.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

DATA_PATH = Path('../data/raw/your_file.csv')
TARGET = 'value'
```

```python
df = pd.read_csv(DATA_PATH)
df.head()
```

```python
df[TARGET].describe()
```

```python
ax = df[TARGET].plot(figsize=(12, 4), title=f'{TARGET} over time')
ax.set_xlabel('Step')
ax.set_ylabel(TARGET)
plt.show()
```
